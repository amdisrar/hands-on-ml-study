## What Is Machine Learning?

### Concept

**Machine Learning (ML)** is a way of programming computers to learn patterns from data instead of explicitly writing every rule for a task.

In traditional programming, humans define the rules. In machine learning, a learning algorithm uses data to discover patterns and build a **model** that can make predictions.

### Example: Spam Email Detection

Consider a program that identifies spam emails.

**Traditional programming:** We manually write rules to detect spam, such as checking for suspicious words or phrases. When spammers change their approach, we may need to update those rules ourselves.

**Machine learning:** We provide examples of spam and legitimate emails. The learning algorithm identifies patterns that distinguish them and builds a model that can classify new emails.

The important difference is that ML learns useful patterns from examples rather than relying entirely on manually written detection rules.

Simply storing thousands of emails is not machine learning. The system must learn something useful from the data to improve its ability to perform a task.

### How Do We Know a Computer Is Learning?

A machine learning system can be understood using three elements:

| Element | Meaning | Spam Filter Example |
|---|---|---|
| Task (T) | What the system should do | Identify spam emails |
| Experience (E) | Data used for learning | Examples of spam and legitimate emails |
| Performance (P) | How well it performs the task | Percentage of correctly classified emails |

A program demonstrates learning when its performance on a task improves through experience.

For example, if a spam filter correctly classifies 90 out of 100 test emails, its **accuracy is 90%**.

### Important Terms

| Term | Simple Explanation |
|---|---|
| Training set | Collection of examples used for learning |
| Training instance (sample) | One example from the training set, such as one email |
| Training | The process of learning patterns from data |
| Model | The learned system used to make predictions |
| Accuracy | The proportion of predictions that are correct |

### Key Takeaway

**Machine learning is not just storing or processing data. It is learning useful patterns from experience to perform a task, including making predictions on new data.**


## Why Use Machine Learning?

### Concept

Machine learning is useful when traditional programming becomes too complicated, difficult to maintain, or unable to solve a problem effectively.

Instead of manually writing every rule, we can train a machine learning model to identify useful patterns from data.

### Four Main Reasons to Use Machine Learning

**1. Simplifies Complex Rule-Based Systems**

Traditional programs may require hundreds of manually written rules.

For example, a traditional spam filter needs rules to identify suspicious words, phrases, and email characteristics. Maintaining these rules becomes difficult as spam techniques change.

Machine learning can learn these patterns from examples, often making the system easier to maintain.

**2. Solves Problems That Are Difficult to Program**

Some problems are too complicated to solve using manually written rules.

For example, speech recognition must understand different accents, speaking speeds, languages, and background noises.

Writing rules for every possible variation is impractical. Machine learning can learn speech patterns from large collections of recordings.

**3. Adapts to Changing Environments**

Traditional programs generally require manual updates when their rules become outdated.

Machine learning systems can adapt by learning from updated training data.

For example, when spammers start using new phrases to avoid detection, an ML-based spam filter can be retrained to recognize these changing patterns.

**Important:** Machine learning models do not necessarily adapt automatically. They may require new training data and retraining.

**4. Discovers Hidden Patterns in Data**

Machine learning can identify relationships and patterns in large datasets that humans might overlook.

For example, an ML model trained on spam emails might discover unusual combinations of words that strongly indicate spam.

Discovering useful, previously unknown patterns in large datasets is called **data mining**.

### Practical Example: Network Intrusion Detection

Consider a network intrusion detection system.

**Traditional Signature-Based Detection:**

- Detects known attacks using predefined signatures.
- Can efficiently identify recognized malicious traffic.
- Requires updated signatures to detect new attack patterns.

**Machine Learning-Based Detection:**

- Learns characteristics of normal and malicious network traffic.
- Can identify suspicious patterns that existing signatures might miss.
- May help detect previously unseen attack techniques.

However, ML cannot guarantee detection of unknown attacks. It may also incorrectly classify legitimate network traffic as malicious. These incorrect alerts are called **false positives**.

**Combining signature-based detection with machine learning can provide complementary security capabilities.**

### Important Terms

| Term | Simple Explanation |
|---|---|
| Data mining | Discovering hidden patterns and useful information in large datasets. |
| Retraining | Training a model again using updated data. |
| False positive | Incorrectly identifying something normal as suspicious or malicious. |

### Key Takeaway

Machine learning is especially useful when traditional programming requires too many rules, when problems are difficult to solve manually, when environments change, or when we want to discover hidden patterns in data.

**Machine learning does not always replace traditional programming. Both approaches can work together to solve problems more effectively.**

## Types of Machine Learning Systems

### Concept

Machine learning systems can be grouped in different ways. The book introduces three questions that help us understand these differences:

1. **Training supervision:** Does the system learn from examples with known answers, without known answers, or using a mixture?
2. **Batch versus online learning:** Does it learn from a fixed training process or update its learning as new data arrives?
3. **Instance-based versus model-based learning:** Does it compare new examples with known examples, or use patterns learned in a model?

These categories can overlap. One ML system may fit into several categories at the same time.

For now, we are studying **training supervision**. We will discuss the other two distinctions when we reach them in the book.

### Key Takeaway

**There is no single type of machine learning system.** We classify systems according to how they learn and how they make predictions.

### Supervised Learning

#### Concept

In **supervised learning**, the training data includes the correct answers, called **labels** or **targets**. The algorithm studies examples to learn the relationship between the input information and the correct answers.

After training, it uses what it learned to predict answers for new data.

#### Example: Spam Email Detection

We give the system emails that have already been marked as **Spam** or **Not Spam**.

The algorithm learns patterns from those examples. When a new email arrives, it predicts which category it belongs to.

This is **classification**, because the answer is a category.

#### Classification Versus Regression

Another supervised learning task is **regression**, where the model predicts a number instead of a category.

For example, to predict the price of a used car, we may provide its age, mileage, and brand. We also give the correct selling prices for examples in the training data.

| Task | Example | Predicted answer |
|---|---|---|
| Classification | Is this email spam? | Spam or Not Spam (a category) |
| Regression | What is this car worth? | A price (a number) |

#### Important Terms

| Term | Simple Explanation |
|---|---|
| Label / target | The correct answer provided for a training example |
| Feature | An input detail used for learning, such as a car's mileage |
| Classification | Predicting a category |
| Regression | Predicting a numeric value |

#### Key Takeaway

**Supervised learning learns from examples that include correct answers.** Classification predicts categories, while regression predicts numbers.

### Unsupervised Learning

#### Concept

In **unsupervised learning**, the training data does **not** contain the correct answers or labels. The algorithm looks for useful patterns or structure on its own.

Unlike supervised learning, we do not tell it which answer each example should have.

#### Example: Grouping Customers

Imagine we have records of 5,000 customers, including their purchases and spending habits, but no labels describing customer types.

A **clustering** algorithm may discover groups of customers with similar behavior. We can then examine those groups and decide whether they are useful.

The algorithm finds similarities; it does not automatically know what the groups *mean*.

#### Example: Unusual Network Traffic

An **anomaly detection** system can learn what normal network traffic usually looks like and flag connections that are very different.

However, **unusual does not always mean malicious**. A rare connection could be legitimate, so further investigation may be needed.

#### Important Terms

| Term | Simple Explanation |
|---|---|
| Unlabeled data | Examples that do not include known answers |
| Clustering | Finding groups of similar examples |
| Anomaly detection | Finding examples that are unusual compared with normal patterns |

#### Key Takeaway

**Unsupervised learning discovers patterns in unlabeled data.** It can group similar records or identify unusual ones, but humans still need to interpret what these findings mean.

### Semi-supervised Learning

#### Concept

**Semi-supervised learning** uses a small amount of **labeled data** together with a larger amount of **unlabeled data**.

It is useful when labeling examples takes time or costs money. Some approaches discover groups in unlabeled data and use the known labels to help train a model.

#### Example(s)

**1. Google Photos — Grouping Similar Faces**

Imagine uploading many family photographs without adding anyone's name. Google Photos can group photographs showing the same person by finding similarities between faces. This is the **unsupervised** part.

When we name a person in a group, such as Ali or Sara, we provide a **label**. The system can use that information to help identify the same person in other photographs. This illustrates how unlabeled photos and a few labels can work together.

**2. Customer Risk Records**

Suppose we have **5,000 customer records**, but only **100** have labels: *High Risk* or *Low Risk*. The other **4,900** records are unlabeled.

A clustering algorithm might group similar customers. For example, a cluster of 500 customers contains 4 labeled High Risk customers, 1 labeled Low Risk customer, and 495 unlabeled customers. A simple method could assign the **most common label (High Risk)** to the other 495 records.

**This is an assumption, not a guarantee.** Customers might have similar ages or spending habits but different risk levels. We should not automatically trust the assigned labels. Also, using extra data does not necessarily make training faster or more accurate.

Clustering is one possible approach; **not all semi-supervised learning methods use clustering**.

#### Important Terms

| Term | Simple Explanation |
|---|---|
| Labeled data | Examples that already have the correct answers |
| Unlabeled data | Examples without known answers |
| Clustering | Grouping examples that have similar features |
| Most common label | The label that appears most often among the labeled examples in a group |

#### Key Takeaway

**Semi-supervised learning combines a few labeled examples with many unlabeled examples.** It may help a model learn more, but similarities between records do not guarantee that they have the same label.

### Self-supervised Learning

#### Concept

**Self-supervised learning** automatically creates training targets from data that originally has no human-provided labels.

The model can then learn using these generated targets. The important difference from ordinary supervised learning is **where the target comes from**: the data itself, rather than a person labeling it.

#### Example(s)

**1. Reconstructing Missing Parts of Photographs**

Imagine we have **10,000 unlabeled photographs of animals**. We hide part of each photograph and ask a model to reconstruct it.

- **Input:** A photograph with a part hidden.
- **Target:** The original complete photograph.
- **Training:** The model compares its reconstruction with the original photograph and improves through repeated examples.

Nobody tells the model whether a photograph is of a cat or a dog. While reconstructing images, it can learn useful visual features such as shapes, textures, and facial structures.

**2. Using the Learned Features to Classify Pets**

Suppose our actual goal is to classify photographs as **Cat** or **Dog**. Instead of starting from scratch, we may reuse the model trained on the 10,000 photographs.

We then **fine-tune** it using, for example, **100 labeled photographs**. The model has already learned useful image features, so it mainly needs to learn how those features relate to the new labels. This reuse of learning is called **transfer learning**. It may improve results, but better accuracy is not guaranteed.

#### Important Terms

| Term | Simple Explanation |
|---|---|
| Self-supervised learning | Learning using targets created automatically from the original data |
| Masking | Hiding part of the input, such as pixels in a photograph |
| Training target | The correct answer the model tries to predict; here, the original image |
| Pretraining | Initial training that helps a model learn useful features |
| Fine-tuning | Training a pretrained model further for a specific task |
| Transfer learning | Reusing knowledge learned for one task in another task |

#### Key Takeaway

**Self-supervised learning creates training targets from unlabeled data itself.** Unlike clustering, it trains with those generated targets. The knowledge learned can later be reused for another task.

### Reinforcement Learning

#### Concept

**Reinforcement Learning (RL)** is a type of machine learning in which an agent learns by interacting with its environment and receiving rewards or penalties for its actions.

The agent learns a strategy, called a **policy**, to choose actions that maximize rewards over time.

#### Example(s)

**1. Teaching a Robot to Walk**

Imagine a robot learning to walk without being given exact instructions.

It tries different movements and receives rewards when it walks successfully and penalties when it falls.

Over many attempts, the robot learns which movements work better.

If the robot moves from a smooth floor to a slippery floor, it may need to continue learning and improve its policy for the new environment.

**2. AlphaGo**

AlphaGo is an AI system developed to play the board game Go.

It learned by studying games and playing many games against itself. Through this process, it developed a strong policy for selecting moves.

#### Important Terms

| Term | Simple Explanation |
|---|---|
| Agent | The system that learns and takes actions |
| Environment | The surroundings or system the agent interacts with |
| Action | A decision or movement made by the agent |
| Reward | Feedback indicating how desirable an outcome is |
| Penalty | Negative feedback, often represented as a negative reward |
| Policy | The strategy used by the agent to select actions |

#### Key Takeaway

**Reinforcement learning is learning through experience, actions, and feedback.** The agent learns a policy that aims to maximize rewards over time, rather than simply choosing actions that give immediate rewards.

### Batch Versus Online Learning

#### Concept

Machine learning systems can be classified by **how they learn from training data**: training a new model on the complete available dataset or updating an existing model in smaller steps.

**Batch Learning (Offline Learning)**

In batch learning, a model is trained on the complete available training dataset. After deployment, it makes predictions but does not keep learning. To learn from new data, we train a new model on an updated dataset and replace the old model.

**Online Learning (Incremental Learning)**

In online learning, a model learns incrementally from individual training examples or small groups called **mini-batches**. It can update its knowledge without training again from scratch on the full dataset.

**Comparison: Batch Learning vs. Online Learning**

| Feature | Batch Learning | Online Learning |
|---|---|---|
| Training method | Train a new model on the complete dataset | Update a model incrementally |
| New data | Retrain a new model | Update using new examples |
| Adaptation | Usually less frequent | Can be frequent |
| Computing resources | Full retraining may be expensive | Small updates usually need fewer resources |
| Large datasets | Full training may require substantial resources | Can process data in small portions |
| Main risk | Model may become outdated | Bad new data may harm the model |
| Often useful when | Data is fairly stable or updates are periodic | Data changes often or cannot fit in memory |

**Important:** Online learning does **not** require an internet connection or live data. It means the algorithm supports incremental updates, including when training offline.

#### Example(s)

**1. Spam Detection — Batch Learning**

A spam detection model is trained on 100,000 emails and deployed. When new spam patterns appear, the deployed model does not learn them automatically. We train a new model using the old and new training data, then deploy it.

**2. Network Intrusion Detection — Online Learning**

An intrusion detection model can learn incrementally from new **labeled** traffic records to adapt to changing attack patterns. However, incorrect labels or other bad training data may harm performance, so the learning process needs monitoring.

**3. Large Dataset — Out-of-core Learning**

Suppose we have **200 GB** of historical network traffic data but only **16 GB of RAM**. An algorithm that supports incremental learning can read small portions from disk and update the model one portion at a time.

This is **out-of-core learning**. It can happen entirely offline.

#### Important Terms

| Term | Simple Explanation |
|---|---|
| Batch Learning | Training a new model using the full available training dataset |
| Offline Learning | Training separately before deploying a model |
| Online Learning | Updating a model incrementally as training examples are processed |
| Mini-batch | A small group of training examples processed together |
| Out-of-core Learning | Training on data too large for memory by processing small portions |
| Data Drift / Model Rot | Changes in the data or real world can make a model less effective over time |
| Learning Rate | Controls how strongly a model adjusts during a training update |
| Catastrophic Forgetting | When learning new information causes a model to lose previously learned knowledge |
| Inertia | Resistance to rapid change; a low learning rate generally means more inertia |
| Noise | Errors or random variation in training data |
| Outlier | A data point very different from most other examples |

#### Key Takeaway

**Batch learning trains a new model on a complete dataset, while online learning updates a model incrementally.** Online learning can adapt faster and handle very large datasets, but it needs monitoring because poor data or overly large updates may reduce performance.

## Instance-Based Versus Model-Based Learning

### Concept

Machine learning systems can also be classified by **how they generalize**: how they use training examples to make predictions about new, unseen examples.

There are two main approaches: **instance-based learning** and **model-based learning**. Good performance on training examples is not enough; the goal is to make useful predictions on new data.

### Instance-Based Learning

#### Concept

In **instance-based learning**, the system remembers training examples and makes predictions by comparing new examples with stored ones (or a subset of them).

Instead of requiring an exact match, it uses a **similarity measure** to find examples that resemble the new observation. Learning from these similarities allows it to **generalize** to new cases.

#### Example: Email Spam Detection

Suppose we have previously labeled emails marked **Spam** or **Not Spam**. Simply remembering identical spam emails would miss new messages with slightly different wording.

A simple similarity measure could count the number of words two emails have in common. If a new email resembles known spam emails, the system may classify it as **Spam**, even though the exact message was not in the training data.

#### Diagram: Using Nearby Examples

![Original diagram of instance-based classification with three nearest neighbors](../../figures/instance-based-learning.svg)

*Original illustration: the new instance is surrounded by stored examples from two classes. Two of its three nearest neighbors are triangles, so a majority-vote classifier predicts **Triangle**.*

#### Intuition: Fruit Classification

Imagine stored examples of apples and oranges, described by features such as weight and size. We can compare a new fruit with the stored fruits to find the closest examples.

With **k-nearest neighbors (k-NN)**, `k` is the number of nearby examples considered. If `k = 5` and four of the closest fruits are apples, a majority-vote classifier predicts **Apple**. If `k = 1` and the single closest fruit is an orange, it predicts **Orange**.

This fruit example and the discussion of `k` are additional learning illustrations; the book uses a general diagram with three nearest neighbors.

#### Advantages and Limitations

| Advantages | Limitations |
|---|---|
| Often useful for small datasets | May require storing and deploying the training examples |
| Can be useful when data keeps changing | Searching for similar instances can make predictions slow |
| Generalizes without requiring exact matches | Does not scale well to very large datasets |
| | Similarity may work poorly with high-dimensional data, such as images |

#### Important Terms

| Term | Simple Explanation |
|---|---|
| Instance | One example or observation in a dataset |
| Generalization | Making useful predictions about examples not seen during training |
| Similarity Measure | A method for comparing how alike two examples are |
| Nearest Neighbor | A stored example that is closest or most similar to a new example |
| k-NN | A method that predicts using the `k` nearest stored examples |

#### Key Takeaway

**Instance-based learning remembers training examples and generalizes by comparing new examples with similar stored ones.** It can work well for small or changing datasets, but storage, prediction speed, and high-dimensional data can be challenges.

### Model-Based Learning and a Typical Machine Learning Workflow

#### Concept

In **model-based learning**, the system learns a general relationship from training data and represents it using a model. The trained model then uses this relationship to make predictions about new, unseen examples.

Unlike instance-based learning, which compares new examples with stored examples, model-based learning uses a learned model to make predictions.

#### Example: Does Money Make People Happier?

Géron uses GDP per capita and life-satisfaction data from different countries to demonstrate model-based learning.

The data shows an approximate upward trend: countries with higher GDP per capita generally tend to have higher life-satisfaction scores in the example.

We can represent this relationship using a **linear model**, which is a straight-line equation.

#### Mathematics: A Simple Linear Model

The book introduces:

$$
y = \theta_0 + \theta_1 x
$$

Here, $y$ represents predicted life satisfaction and $x$ represents GDP per capita. This has the same structure as the familiar equation:

$$
y = mx + c
$$

| Symbol | Simple Explanation |
|---|---|
| $y$ | Predicted life-satisfaction score |
| $x$ | GDP per capita (input) |
| $\theta_0$ | Intercept — the prediction when $x=0$ |
| $\theta_1$ | Slope — how much the prediction changes for each unit increase in $x$ |

The values $\theta_0$ and $\theta_1$ are called **model parameters**. They determine the position and slope of the line.

After training the model, Géron's example obtains approximately:

$$
\theta_0 = 3.75
$$

$$
\theta_1 = 6.78 \times 10^{-5} = 0.0000678
$$

Therefore:

$$
y = 3.75 + 0.0000678x
$$

For a country with GDP per capita of $40,000:

$$
y = 3.75 + (0.0000678 \times 40,000)
$$

$$
y = 3.75 + 2.712 = 6.462
$$

**Predicted life satisfaction: approximately 6.46.**

This is a prediction, not necessarily the country's actual life-satisfaction score.

#### Model Training and Cost Function

Before making predictions, the model must learn suitable parameter values from the training data.

A **performance measure** tells us how well the model performs.

- A **utility function** measures how good the model is. Higher is better.
- A **cost function** (also called a loss function) measures how bad the model is. Lower is better.

A common cost function for linear regression is **Mean Squared Error (MSE)**.

$$
MSE = \frac{\text{Sum of Squared Errors}}{\text{Number of Examples}}
$$

For example, if the actual value is 6 and the prediction is 8:

$$
\text{Error} = 8 - 6 = 2
$$

$$
\text{Squared Error} = 2^2 = 4
$$

MSE averages the squared errors across all predictions. Squaring prevents positive and negative errors from canceling and gives larger errors greater weight.

**Training** means finding model parameters that minimize the cost function on the training data.

A lower training cost means a better fit to the training data, but it does not guarantee better predictions on unseen data.

#### Typical Machine Learning Workflow

![Original model-based learning workflow diagram](../../figures/model-based-learning-workflow.svg)

The workflow has five main steps:

1. **Prepare data:** Collect and organize inputs and known answers.
2. **Select a model:** Choose the type of mathematical relationship, such as linear regression.
3. **Define a performance measure:** Decide how to evaluate the model's predictions.
4. **Train the model:** Find parameter values that minimize the cost.
5. **Make predictions:** Use the trained model with new input data. This is called **inference**.

#### Important Terms

| Term | Simple Explanation |

#### Key Takeaway

**Model-based learning builds a model from training data, learns its parameters by minimizing a cost function, and uses the trained model to make predictions on new examples.**

The real goal is not simply to fit training data well, but to **generalize well to unseen data**.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Load the dataset
data_root = "https://github.com/ageron/data/raw/main/"
lifesat = pd.read_csv(data_root + "lifesat/lifesat.csv")

# Separate inputs and known answers
X = lifesat[["GDP per capita (USD)"]].values
y = lifesat[["Life satisfaction"]].values

# Select a linear model and train it
model = LinearRegression()
model.fit(X, y)

# Examine the learned parameters
print("Intercept:", model.intercept_)
print("Slope:", model.coef_)

# Predict life satisfaction for Puerto Rico
X_new = [[33_442.8]]
prediction = model.predict(X_new)
print("Predicted life satisfaction:", prediction)
```
